```python
import numpy as np 
import pandas as pd 
```

```python
def clean_model(x):
    """Clean the phone model name by removing tags like 4G, 5G, and Dual Sim."""
    
    x = x.split("(")[0]
    
    x = x.replace("5G", "").replace("4G", "").replace("Dual Sim", "")
    
    words = x.strip().split()
    if len(words) > 0 and len(words[-1]) == 4 and words[-1].isdigit():
        words.pop()
        
    return " ".join(words).strip()

df["Model"] = df["model"].apply(clean_model)
```

```python
# Dropping the model column 
df.drop('model', axis=1, inplace=True)
```

```python
df['price'].sample(10)
```

```python
def clean_price(x):
  """Clean the price of phone by replacing '₹'."""
  x = x.strip().replace("₹", "").replace(",", "")
  return x

df['Price'] = df['price'].apply(clean_price)
```

```python
# Dropping price column 
df.drop('price', axis=1, inplace=True)
```

```python
# change the data type of Price column to float64
df['Price'] = df["Price"].astype('float64')
```

```python
def clean_sim_ds(x):
  """Cleans the sim of phone into Dual and Single"""
  if "Dual" in x:
    return 1
  elif "Single" in x:
    return 0
  else:
    return np.nan
  
df["Sim"] = df["sim"].apply(clean_sim_ds)
```

```python
# 5G column - yes or no
def clean_sim_5G(x):
  """Clean the sim column into 5G - Yes or No"""
  if "5G" in x:
    return 1
  else:
    return 0

df['5G'] = df['sim'].apply(clean_sim_5G)
```

```python
df['5G'].dtypes
```

```python
def clean_sim_irb(x):
  """Cleans the sim into IR column - Yes and No"""
  if "IR" in x:
    return 1
  else:
    return 0
  
df['IrBlaster'] = df["sim"].apply(clean_sim_irb)
```

```python
def clean_sim_nfc(x):
  """Cleans the sim column into NFC column - Yes and No"""
  if "NFC" in x:
    return 1
  else:
    return 0
df["NFC"] = df["sim"].apply(clean_sim_nfc)
```

```python
# drop sim column 
df.drop('sim', axis=1, inplace=True)
```

```python
def clean_processor_speed(x):
  """Extract the numeric processor speed (GHz) from the processor string."""
  if "GHz" in x:
    return x.strip().split(",")[-1].strip().split(" ")[0].replace("GHz", "")
  else:
    return np.nan
  
df['Speed'] = df['processor'].apply(clean_processor_speed)
```

```python
# change the datatype of Speed column to float
df["Speed"] = df["Speed"].astype("float64")
```

```python
def clean_processor_core(x):
  """Extract the core count (e.g., Octa, Quad) from the processor string."""

  if "Core" in x:
      x = x.strip().split(",")
      c = " ".join(x).strip().split(" ")
      if "Core" in c:
        index = c.index("Core")
        return c[index - 1]
      else:
        return np.nan
  else:
      return np.nan

df["Core"] = df["processor"].apply(clean_processor_core)
```

```python
df['Core'].unique()
```

```python
# map the Core column as per the unique values
core_map = {
    'Octa': 8,
    'Hexa': 6,
    'Quad': 4,
    'Dual': 2,
    'Single': 1
}

df['Core'] = df['Core'].map(core_map)
```

```python
def clean_processor_pn(x):
    """Extract processor brand/series from processor column."""
    
    if pd.isna(x):
        return np.nan

    c = str(x).split(",")[0].strip().lower()

    if "snapdragon" in c or "sanpdragon" in c or "qualcomm" in c:
        return "snapdragon"
    elif "dimensity" in c:
        return "dimensity"
    elif "helio" in c:
        return "helio"
    elif "unisoc" in c or "spreadtrum" in c or "sc9863a" in c or "sc6531e" in c or "tiger" in c:
        return "unisoc"
    elif "exynos" in c or "samsung" in c:
        return "exynos"
    elif "bionic" in c or "a13" in c or "a14" in c or "a15" in c or "a16" in c or "fusion" in c:
        return "bionic"
    elif "google" in c or "tensor" in c:
        return "google"
    elif "kirin" in c:
        return "kirin"
    else:
        return np.nan

df["ProcessorName"] = df["processor"].apply(clean_processor_pn)
```

```python
df[['processor', 'ProcessorName']].sample(10)
```

```python
# drop the processor column 
df.drop('processor', axis=1, inplace=True)
```

```python
def clean_ram_ram(x):
    """Normalize RAM to a numeric GB value directly using float."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "GB RAM" in x:
        ram = x.split(",")[0].replace("GB RAM", "").replace("GB", "").strip()
        return float(ram)
            
    elif "MB RAM" in x:
        ram = x.split(",")[0].replace("MB RAM", "").replace("MB", "").strip()
        return round(float(ram) / 1024, 4)

    else:
        return np.nan

df['Ram'] = df['ram'].apply(clean_ram_ram)
```
