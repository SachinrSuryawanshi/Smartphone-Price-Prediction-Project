```python
def clean_model(x):
    """Clean the phone model name by removing tags like 4G, 5G, and Dual Sim."""
    
    x = x.split("(")[0]
    
    x = x.replace("5G", "").replace("4G", "").replace("Dual Sim", "")
    
    words = x.strip().split()
    if len(words) > 0 and len(words[-1]) == 4 and words[-1].isdigit():
        words.pop()
        
    return " ".join(words).strip()

df["Model"] = df["model"].apply(clean_model)
```

```python
def clean_price(x):
  """Clean the price of phone by replacing '₹'."""
  x = x.strip().replace("₹", "").replace(",", "")
  return x

df['Price'] = df['price'].apply(clean_price)
```

```python
def clean_sim_ds(x):
  """Cleans the sim of phone into Dual and Single"""
  if "Dual" in x:
    return 1
  elif "Single" in x:
    return 0
  else:
    return np.nan
  
df["Sim"] = df["sim"].apply(clean_sim_ds)
```

```python
# 5G column - yes or no
def clean_sim_5G(x):
  """Clean the sim column into 5G - Yes or No"""
  if "5G" in x:
    return 1
  else:
    return 0

df['5G'] = df['sim'].apply(clean_sim_5G)
```

```python
def clean_sim_irb(x):
  """Cleans the sim into IR column - Yes and No"""
  if "IR" in x:
    return 1
  else:
    return 0
  
df['IrBlaster'] = df["sim"].apply(clean_sim_irb)
```

```python
def clean_sim_nfc(x):
  """Cleans the sim column into NFC column - Yes and No"""
  if "NFC" in x:
    return 1
  else:
    return 0
df["NFC"] = df["sim"].apply(clean_sim_nfc)
```

```python
# drop sim column 
df.drop('sim', axis=1, inplace=True)
```

```python
def clean_processor_speed(x):
  """Extract the numeric processor speed (GHz) from the processor string."""
  if "GHz" in x:
    return x.strip().split(",")[-1].strip().split(" ")[0].replace("GHz", "")
  else:
    return np.nan
  
df['Speed'] = df['processor'].apply(clean_processor_speed)
```

```python
# change the datatype of Speed column to float
df["Speed"] = df["Speed"].astype("float64")
```

```python
def clean_processor_core(x):
  """Extract the core count (e.g., Octa, Quad) from the processor string."""

  if "Core" in x:
      x = x.strip().split(",")
      c = " ".join(x).strip().split(" ")
      if "Core" in c:
        index = c.index("Core")
        return c[index - 1]
      else:
        return np.nan
  else:
      return np.nan

df["Core"] = df["processor"].apply(clean_processor_core)
```

```python
df['Core'].unique()
```

```python
# map the Core column as per the unique values
core_map = {
    'Octa': 8,
    'Hexa': 6,
    'Quad': 4,
    'Dual': 2,
    'Single': 1
}

df['Core'] = df['Core'].map(core_map)
```

```python
def clean_processor_pn(x):
    """Extract processor brand/series from processor column."""
    
    if pd.isna(x):
        return np.nan

    c = str(x).split(",")[0].strip().lower()

    if "snapdragon" in c or "sanpdragon" in c or "qualcomm" in c:
        return "snapdragon"
    elif "dimensity" in c:
        return "dimensity"
    elif "helio" in c:
        return "helio"
    elif "unisoc" in c or "spreadtrum" in c or "sc9863a" in c or "sc6531e" in c or "tiger" in c:
        return "unisoc"
    elif "exynos" in c or "samsung" in c:
        return "exynos"
    elif "bionic" in c or "a13" in c or "a14" in c or "a15" in c or "a16" in c or "fusion" in c:
        return "bionic"
    elif "google" in c or "tensor" in c:
        return "google"
    elif "kirin" in c:
        return "kirin"
    else:
        return np.nan

df["ProcessorName"] = df["processor"].apply(clean_processor_pn)
```

```python
def clean_ram_ram(x):
    """Normalize RAM to a numeric GB value directly using float."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "GB RAM" in x:
        ram = x.split(",")[0].replace("GB RAM", "").replace("GB", "").strip()
        return float(ram)
            
    elif "MB RAM" in x:
        ram = x.split(",")[0].replace("MB RAM", "").replace("MB", "").strip()
        return round(float(ram) / 1024, 4)

    else:
        return np.nan

df['Ram'] = df['ram'].apply(clean_ram_ram)
```

```python
df[['ram','Ram']].sample(10)
```

```python
def clean_ram_rom(x):
    """Extract internal storage (ROM) size and normalize to numeric GB."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()

    if "TB inbuilt" in x:
        rom = x.split(",")[-1].replace("TB inbuilt", "").replace("TB", "").strip()
        return float(rom) * 1024

    elif "GB inbuilt" in x:
        rom = x.split(",")[-1].replace("GB inbuilt", "").replace("GB", "").strip()
        return float(rom)

    elif "MB inbuilt" in x:
        rom = x.split(",")[-1].replace("MB inbuilt", "").replace("MB", "").strip()
        return round(float(rom) / 1024, 4)

    return np.nan

df["Rom"] = df["ram"].apply(clean_ram_rom)
```

```python
def clean_battery_capacity(x):
    """Extract the numeric battery capacity (mAh) from the battery string."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "mAh" in x:
        # Take everything before 'mAh', strip whitespace, and convert to float
        cap = x.split("mAh")[0].strip()
        try:
            return float(cap)
        except ValueError:
            return np.nan
            
    return np.nan

df["Capacity"] = df["battery"].apply(clean_battery_capacity)
```

```python
df['Capacity']
```

```python
def clean_battery_watt(x):
    """Extract the fast-charging wattage (e.g., 33, 65) from the battery string."""
    if pd.isna(x):
        return np.nan
    
    c = str(x).split()
    
    if "Fast" in c:
        index = c.index("Fast")
        prev_word = c[index - 1]
        
        if prev_word == "with" or not prev_word.endswith("W"): # if the word right before 'Fast' is 'with', wattage wasn't specified
            return np.nan
        
        return float(prev_word.replace("W", "").strip()) # remove 'W' and convert to numeric float
        
    return np.nan

df["Watt"] = df["battery"].apply(clean_battery_watt)
```

```python
df['Watt']
```

```python
def clean_battery_fc(x):
  """Categorize whether the phone supports fast charging (Yes/No) ---> (1/0)."""

  if "Fast" in x:
    return 1
  else:
    return 0
  
df["FastCharging"] = df["battery"].apply(clean_battery_fc)
```

```python
def clean_display_inches(x):
  """Extract the display size in inches from the screen specification string."""

  if "inches" in x:
    inche = x.strip().split(",")[0].replace("inches", "")
    return float(inche)
  else:
    return np.nan
df["Inches"] = df["display"].apply(clean_display_inches)
```

```python
df['Inches']
```

```python
def clean_display_screen_resol(x):
    """Extract the horizontal pixel count (X-axis resolution) from the display string."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "px" in x and "x" in x:  # check if resolution pattern exists (e.g. '1080 x 2400 px')

        res = x.split("px")[0]
        
        if "x" in res:
            x_res = res.split("x")[0].split(",")[-1].strip()
            try:
                return float(x_res)
            except ValueError:
                return np.nan
                
    return np.nan

df["Xaxis"] = df["display"].apply(clean_display_screen_resol)
```

```python
df['Xaxis']
```

```python
def clean_display_resolu_y(x):
    """Extract the vertical pixel count (Y-axis resolution) from the display string."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "px" in x and "x" in x:
        res = x.split("px")[0]
        
        if "x" in res:
            y_res = res.split("x")[1].strip()
            try:
                return float(y_res)
            except ValueError:
                return np.nan
                
    return np.nan

df["Yaxis"] = df["display"].apply(clean_display_resolu_y)
```

```python
df['Yaxis']
```

```python
def clean_display_hz(x):
    """Extract the screen refresh rate (e.g., 60, 90, 120) in Hz from the display string."""
    if pd.isna(x):
        return np.nan
    
    c = str(x).replace(",", " ").split()
    
    if "Hz" in c:
        index = c.index("Hz")
        try:
            return float(c[index - 1])
        except ValueError:
            return np.nan
        
    return np.nan

df["Hz"] = df["display"].apply(clean_display_hz)
```

```python
df['Hz'].dtypes
```

```python
def clean_display_notches(x):
    """Identify the notch type from the display string."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).lower()
    
    if "dual punch hole" in x:
        return "Dual Punch Hole"
    elif "punch hole" in x:
        return "Punch Hole"
    elif "water drop" in x or "waterdrop" in x:
        return "Water Drop Notch"
    elif "small notch" in x:
        return "Small Notch"
    elif "large notch" in x:
        return "Large Notch"
    elif "notch" in x:
        return "Notch"
    
    return np.nan

df["Notches"] = df["display"].apply(clean_display_notches)
```

```python
def clean_primary_rear_mp(x):
    """Extract the main/first rear camera megapixel value."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "No Rear Camera" in x:
        return 0.0
    
    if "MP" in x:
        val = x.split("MP")[0].strip()
        val = val.split("+")[0].strip() # Handle cases like '50 MP + 8 MP' -> val is '50'
        try:
            return float(val)
        except ValueError:
            return np.nan
            
    return np.nan

df["RearMP"] = df["camera"].apply(clean_primary_rear_mp)
```

```python
df['RearMP'].dtypes
```

```python
def clean_front_mp(x):
    """Extract the front camera megapixel value."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "Front Camera" in x and "&" in x:
        front = x.split("&")[1].split("MP")[0].strip()
        try:
            return float(front)
        except ValueError:
            return np.nan
            
    elif "No Front Camera" in x or "Rear Camera" in x:
        return 0.0
        
    return np.nan

df["FrontMP"] = df["camera"].apply(clean_front_mp)
```

```python
df['FrontMP'].dtypes
```

```python
def clean_rear_camera_count(x):
    """Extract the number of rear camera lenses."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).lower()
    
    if "no rear camera" in x:
        return 0
    elif "quad" in x:
        return 4
    elif "triple" in x:
        return 3
    elif "dual" in x:
        return 2
    elif "rear camera" in x or "mp" in x:
        return 1
        
    return np.nan

df["RearCameraCount"] = df["camera"].apply(clean_rear_camera_count)
```

```python
def clean_card_support(x):
    """Return 1 for supported, 0 for not supported, and np.nan for missing/shifted data."""
    if pd.isna(x):
        return np.nan
    
    x = str(x)
    
    if "Memory Card" not in x and "card" not in x.lower():
        return np.nan
    
    if "Not Supported" in x or "No" in x:
        return 0
    
    elif "Supported" in x or "Hybrid" in x:
        return 1
        
    return np.nan

df["HasMemoryCard"] = df["card"].apply(clean_card_support)
```

```python
df['HasMemoryCard'].dtypes
```

```python
def clean_card_hybrid(x):
    """Flag whether the memory card slot is a Hybrid slot (1 = Yes, 0 = Dedicated/No, NaN = Missing/Shifted)."""
    if pd.isna(x):
        return np.nan
    
    x = str(x)
    
    if "Memory Card" not in x and "card" not in x.lower():
        return np.nan
    
    if "Hybrid" in x:
        return 1
    else:
        return 0

df["IsHybrid"] = df["card"].apply(clean_card_hybrid)
```

```python
df['IsHybrid'].isnull().sum()
```

```python
def clean_card_extrasupport(x):
    """Extract expandable storage capacity in GB (e.g., 512, 1024, 0)."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).strip()
    
    if "Not Supported" in x:
        return 0.0
    
    if "upto" in x:
        cap = x.split("upto")[-1].strip()
        
        if "TB" in cap:
            val = cap.replace("TB", "").strip()
            try:
                return float(val) * 1024
            except ValueError:
                return np.nan
                
        elif "GB" in cap:
            val = cap.replace("GB", "").strip()
            try:
                return float(val)
            except ValueError:
                return np.nan
                
    return np.nan

df["ExtraSupport"] = df["card"].apply(clean_card_extrasupport)
```

```python
def clean_os(x):
    """Extract operating system family from the OS column."""
    if pd.isna(x):
        return np.nan
    
    x = str(x).lower()
    
    if "android" in x or "emui" in x or "pragati" in x or "miui" in x or "oxygenos" in x:
        return "android"
    elif "ios" in x:
        return "ios"
    elif "harmonyos" in x or "hongmeng" in x:
        return "harmonyos"
    elif "kaios" in x:
        return "kaios"
    elif "rtos" in x or "symbian" in x or "java" in x:
        return "other"
    else:
        return np.nan

df["OS"] = df["os"].apply(clean_os)
```
