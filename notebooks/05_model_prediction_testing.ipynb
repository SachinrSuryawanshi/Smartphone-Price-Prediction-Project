Load Trained Model and Make Prediction

```python
import pandas as pd
import numpy as np
import joblib
```

```python
MODEL_PATH = "D:/Smartphone-Price-Prediction-Project/Smartphone-Price-Prediction-Project/models/smartphone_price_model.joblib"
```

```python
# load the trained model
pipeline = joblib.load(MODEL_PATH)
```

```python
# test the model with a sample input
sample_input = pd.DataFrame({
    "Brand": ["samsung"],
    "OS": ["android"],
    "Notches": ["punch hole"],
    "ProcessorName": ["snapdragon"],
    "rating": [85.0],
    "Speed": [2.8],         
    "Core": [8.0],        
    "Ram": [8.0],          
    "Rom": [256.0],         
    "Inches": [6.7],
    "Xaxis": [1080.0],
    "Yaxis": [2400.0],
    "Hz": [90.0],           
    "Capacity": [5000.0],   
    "Watt": [45.0],          
    "RearCamera": [1.0],    
    "RearMP": [64.0],        
    "FrontMP": [32.0],       
    "RearCameraCount": [3.0],
    "Sim": [1.0],         
    "5G": [1],
    "NFC": [1],
    "IrBlaster": [0],
    "FastCharging": [1],
    "HasMemoryCard": [1],   
    "IsHybrid": [0],         
    "ExtraSupport": [512.0]    
})
```

```python
display(sample_input)
```

```python
prediction_log = pipeline.predict(sample_input)
print(f"Predicted log price: {prediction_log[0]}")
```

```python
# Reverse the log transformation to get actual Rupees
predicted_price = np.expm1(prediction_log[0])
print(f"Predicted Price: ₹{predicted_price:.2f}")
```
