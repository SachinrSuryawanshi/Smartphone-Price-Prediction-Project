```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('D:/Smartphone-Price-Prediction-Project/Smartphone-Price-Prediction-Project/data/processed/cleaned_smartphones.csv')
```

```python
# rating column 
sns.histplot(data = df, x = "rating", kde=True, bins=25, color="darkorange", edgecolor="black")

plt.title("Distribution of Smartphone Ratings")
plt.xlabel("Rating (Score out of 100)")
plt.ylabel("Number of Smartphones")
plt.tight_layout()
plt.show()
```

```python
# Price column
sns.histplot(data=df, x = 'Price', kde=True, bins=35, color="royalblue")

plt.title("Distribution of Price (Right-Skewed)")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# Top 15 smartphone model counts
top_15_brands = df["Model"].value_counts().head(15)

sns.barplot(x=top_15_brands.values, y=top_15_brands.index, palette="Blues_r")
plt.title("Top 15 Smartphone Brands by Model Count")
plt.xlabel("Number of Models")
plt.ylabel("Brand")
plt.show()
```

```python
# top 7 models price distribution
top_7_brands = df["Model"].value_counts().head(7).index
brand_subset = df[df["Model"].isin(top_7_brands)]

plt.figure(figsize=(9, 5))
sns.boxplot(data=brand_subset, x="Model", y="Price", palette="Set3")
plt.yscale("log")
plt.title("Log-Scaled Price Distribution Across Top 7 Brands")
plt.xlabel("Brand")
plt.xticks(rotation=10)
plt.ylabel("Price (Log Scale)")
plt.show()
```

```python
# Price vs Ram column
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Ram", y="Price", hue="5G")
plt.yscale("log")
plt.title("Price vs RAM Capacity (Colored by 5G Support)")
plt.xlabel("RAM (GB)")
plt.ylabel("Price (Log Scale)")
plt.show()
```

```python
plt.figure(figsize=(14, 10))
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

# Focus heatmap around target columns
sns.heatmap(corr, cmap="vlag", vmin=-1, vmax=1, annot=False, linewidths=0.3)
plt.title("Correlation Matrix Across All Cleaned Numeric Features", fontsize=14, fontweight="bold")
plt.show()

print("\n--- Features Most Correlated with Price ---")
print(corr["Price"].dropna().sort_values(ascending=False).to_frame().head(10))

print("\n--- Features Most Correlated with Rating ---")
print(corr["rating"].dropna().sort_values(ascending=False).to_frame().head(10))
```
