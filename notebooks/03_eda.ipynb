```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
# rating column 
sns.histplot(data = df, x = "rating", kde=True, bins=25, color="darkorange", edgecolor="black")

plt.title("Distribution of Smartphone Ratings")
plt.xlabel("Rating (Score out of 100)")
plt.ylabel("Number of Smartphones")
plt.tight_layout()
plt.show()
```

```python
# Price column
sns.histplot(data=df, x = 'Price', kde=True, bins=35, color="royalblue")

plt.title("Distribution of Price (Right-Skewed)")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# Top 15 smartphone model counts
top_15_brands = df["Model"].value_counts().head(15)

sns.barplot(x=top_15_brands.values, y=top_15_brands.index, palette="Blues_r")
plt.title("Top 15 Smartphone Brands by Model Count")
plt.xlabel("Number of Models")
plt.ylabel("Brand")
plt.show()
```

```python
# top 7 models price distribution
top_7_brands = df["Model"].value_counts().head(7).index
brand_subset = df[df["Model"].isin(top_7_brands)]

plt.figure(figsize=(9, 5))
sns.boxplot(data=brand_subset, x="Model", y="Price", palette="Set3")
plt.yscale("log")
plt.title("Log-Scaled Price Distribution Across Top 7 Brands")
plt.xlabel("Brand")
plt.xticks(rotation=10)
plt.ylabel("Price (Log Scale)")
plt.show()
```

```python
# combined resolution label
res_df = df.dropna(subset=["Xaxis", "Yaxis"]).copy()
res_df["Resolution"] = res_df["Xaxis"].astype(int).astype(str) + " x " + res_df["Yaxis"].astype(int).astype(str)

# selecting top 6 most common resolutions
top_res_names = res_df["Resolution"].value_counts().head(6).index

sns.boxplot(
    data=res_df[res_df["Resolution"].isin(top_res_names)],
    x="Resolution",
    y="Price",
    palette="Set2"
)
plt.yscale("log")
plt.title("Price Distribution by Top Screen Resolutions", fontsize=13, fontweight="bold")
plt.xlabel("Screen Resolution (px)", fontsize=11)
plt.ylabel("Price (log scale)", fontsize=11)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
